# 02 — Empirical Orthogonal Function (EOF) Analysis
Fits PCA decomposition across 15 standard depth levels (0–1000m) and computes cumulative explained variance.

```python
import numpy as np
import matplotlib.pyplot as plt
from src.data.eof import ProfileEOF
from src.serving.inference import DEPTHS

# Simulate realistic profiles
rng = np.random.default_rng(42)
depths = np.array(DEPTHS)
profiles = np.zeros((1000, len(depths)))
for i in range(1000):
    sst = 28.0 + rng.normal(0, 1.5)
    profiles[i] = (sst - 4.0) * np.exp(-depths / 240.0) + 4.0 + rng.normal(0, 0.2, len(depths))

eof = ProfileEOF(n_modes=15).fit(profiles)
cumvar = np.cumsum(eof.pca.explained_variance_ratio_)

plt.figure(figsize=(8, 4))
plt.plot(range(1, len(cumvar) + 1), cumvar * 100, marker='o', color='#00d4ff')
plt.axhline(99, color='r', linestyle='--', label='99% Variance')
plt.xlabel('EOF Modes')
plt.ylabel('Cumulative Variance Explained (%)')
plt.title('EOF Spectrum for North Indian Ocean Subsurface Profiles')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
