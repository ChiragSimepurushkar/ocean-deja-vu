# 01 — Exploratory Data Analysis: Satellite Observations
Inspects spatial distributions, seasonal climatology, and correlation structure across the 15 surface channels over the North Indian Ocean (5–30°N, 45–105°E).

```python
import numpy as np
import matplotlib.pyplot as plt
from src.data.features import coordinate_channels, wind_stress_curl
from src.serving.cache_warmer import generate_mock_field

# Visualize surface SST and MLD
sst = generate_mock_field('temp_0m')
mld = generate_mock_field('mld')

fig, axs = plt.subplots(1, 2, figsize=(14, 5))
im0 = axs[0].imshow(sst, origin='lower', cmap='RdYlBu_r')
axs[0].set_title('Sea Surface Temperature (OSTIA)')
plt.colorbar(im0, ax=axs[0])

im1 = axs[1].imshow(mld, origin='lower', cmap='Blues_r')
axs[1].set_title('Mixed Layer Depth (m)')
plt.colorbar(im1, ax=axs[1])
plt.show()
```
